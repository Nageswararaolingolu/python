### 1. What is a Function in Python?
#### A function is a reusable block of code that performs a specific task.
### 2. Why Do We Need Functions?
#### Functions help us with:
- Code reusability
- Avoiding duplicate code
- Better organization
- Easy debugging
- Easy maintenance
- Breaking a large program into smaller tasks
### 3. Types of Functions in Python
#### At a high level, Python functions are divided into two main types:

- Functions
│
├── 1. Built-in Functions
│
└── 2. User-defined Functions
#### 1. Built-in Functions
These are functions already provided by Python.
Examples:
- print()
- len()
- sum()
- max()
- min()
- type()
- sorted()
- range()
### 4. User-Defined Functions
#### These are functions created by us according to our requirements.
#### - We use the keyword: def

```python
### Arguments: Positional and keyword arguments
def details(name,salary,age):
    print("name:",name)
    print("salary:",salary)
    print("age",age)
```

```python
details("priya",6500,65)
```

```python
details("priya",65,6500)
```

```python
# Keyword arguments
details(name="priya", salary=6500,age=65)
```

```python
details(salary=6500,name="priya",age=65)
```

```python
### Return
def add(a,b):
    return a+b
```

```python
add(10,20)
```

```python
### Multiple positional arguements(*args) : they are defined in tuples
def sales(a,b,c):
    return a+b+c
```

```python
sales(10,20,20)
```

```python
def sales(*values):
    return sum(values)
```

```python
sales(10,20,30,40,50)
```

```python
sales(10,20,30,40,50,60,70,70,80,80,90)
```

```python
#### Multiple Keyword arguments (**kwargs) : they are defined in dictionary
def details(**kwargs):
    print("Name:",kwargs["name"])
    print("Location:", kwargs["location"])
    print("salary:", kwargs["salary"])
```

```python
# name, location, salary, city
details(
    name="priya",
    location="hyd",
    salary=6500
)
```

```python
# Total_sales, avg_sales, min_sales, max_sals, num of orders (tuples)
# salesperson, region, category (dictionayr)
def kpi_summary(*sales,**details):
    total_sales = sum(sales)
    total_orders = len(sales)
    avg_sales = total_sales/total_orders
    min_sales = min(sales)
    max_sales = max(sales)

    print("*** KPI SUMMARY *****")
    print("Total Sales:",total_sales)
    print("Avg Order VAlue:", avg_sales)
    print("Num of Orders:", total_orders)
    print("Min sale value:", min_sales)
    print("Max sale value:",max_sales)
    print("\n")
    print("**** SALESPERSON DETAILS ***")
    print("Salesperson Name:",details["name"])
    print("Region:", details["region"])
    print("Category:", details["category"])
    
```

```python
kpi_summary(
    100,200,300,400,500,600,700,800,100,
    name="Ravi", region="Hyd", category="Electronics"
)
```
