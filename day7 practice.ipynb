```python
def kpi_summary(*sales,**details):
    total_sales=sum(sales)
    total_orders=len(sales)
    avg_sales=total_sales/total_orders
    min_sales=min(sales)
    max_sales=max(sales)

    print("****Kpi_Summary***")
    print("total_sales:",total_sales)
    print("total_orders:",total_orders)
    print("min_saales:",min_sales)
    print("max_sales",max_sales)
    print("***sales_person Details***")
    print("sales person name:",details["name"])
    print("region:",details["region"])
    print("category:",details["category"])
```

```python
kpi_summary(
    100,200,300,400,500,600,700,800,900.1000,
    name="siddu",region="Ap",category="electronics"
)
```
